```python
def calcule_Temperature(fille_txt):
  ''' This function take a text fille like input and calcule max ,min,total rain ,highest,lowest of temperature'''
  
  with open(fille_txt, 'r') as file:
        lines = file.readlines()        
  Max = 0
  Min = 0
  Prec = 0
  Count = 0
  rainy_days = 0
  highest = 0
  lowest = 0

  for line in lines :  
    max_Temp=line.split(',')[1]
    min_Temp=line.split(',')[2]
    prec_Temp=line.split(',')[3]
    
    Max += float(max_Temp)
    Min += float(min_Temp)
    Prec += float(prec_Temp)
    Count += 1
    
    
    if float(prec_Temp) > 0:
      rainy_days+=1

    if float(max_Temp) > highest:
          highest = float(max_Temp)

    if float(min_Temp) < lowest:
          lowest = float(min_Temp)


    
  maximum = Max/Count
  minimum = Min/Count
    


  return{"avg_temp_maximum" :maximum,
          'avg_temp_minimum': minimum,
          "Total Rain:" :Prec,
          "Rainy Days:": rainy_days,
          "Highest_Temp:": highest,
          "Lowest_Temp:":lowest}
```

```python
# Function method:  with list comprihchen
def analyse_weather_data(filename):
      """Analyzes weather data from a given file.
    Args: 
        Filename: The name of the weather data file.
    Returns:
        A tuple containing:
            - Average maximum temperature
            - Average minimum temperature
            - Total precipitation
            - Number of rainy days
            - Maximum temperature
            - Minimum temperature"""
      with open(filename,'r') as file:
        lines = file.readlines()
      lines=[line.strip() for line in lines]
      max_temp=[]
      min_temp=[]
      prec=[]
      rainy_days=0
      for i in lines:
              max_temp.append(float(i.split(',')[1]))
              min_temp.append(float(i.split(',')[2]))
              prec.append(float(i.split(',')[3]))
              rainy_days=sum([1 for j in prec if j>0])
      average_max=(sum(max_temp))/len(max_temp)
      average_min=(sum(min_temp))/len(min_temp)
      min_t=min(min_temp)

      max_t=max(max_temp)
      total_prec=sum(prec)
      return {"The average maximum temperature is " : average_max,
              "The average minimum temperature is ": average_min,
              "The total of precipitaion is ":total_prec,
              "The number of rainy days is ": rainy_days,
              "The highest temperature  recorded is ": max_t,
              "The lowest temperature  recorded is ":min_t}
```

```python
analyse_weather_data('weather_data.txt')
```

```python
def weather_statistics(file_path):
    '''This function takes a text file as input and calculates the average maximum temperatures, average minimum temperatures, and total precipitation. It return a dictionary with the calculated statistics.'''
    with open(file_path,'r')as file:
        lines=file.readlines()
    max_temperatures=[]
    min_temperatures=[]
    precipitations=[]
    precipitation_count=0
    for line in lines:
        max_temperature=line.split(',')[1]
        min_temperature=line.split(',')[2]
        precipitation=line.split(',')[3]

        max_temperatures.append(float(max_temperature))
        min_temperatures.append(float(min_temperature))
        precipitations.append(float(precipitation))
    for precipitation in precipitations:
        if float(precipitation) > 0:
            precipitation_count+=1
    avg_max_temp=sum(max_temperatures)/len(max_temperatures)
    avg_min_temp=sum(min_temperatures)/len(min_temperatures)
    total_precipitation=sum(precipitations)
    max_temperature=max(max_temperatures)
    min_temperature=min(min_temperatures)
    return {
        'avg_max_temp': avg_max_temp,
        'avg_min_temp': avg_min_temp,
        'total_precipitation': total_precipitation,
        'max_temperature': max(max_temperatures),
        'min_temperature': min(min_temperatures),
        'precipitation_count': precipitation_count
    }
```

```python
weather_statistics('weather_data.txt')
```
