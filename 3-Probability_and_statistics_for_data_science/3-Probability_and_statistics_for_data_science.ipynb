```python
# read the text file with open
with open('weather_data.txt', 'r') as file:
        lines = file.readlines()        
print(lines)
```

```python
lines[0]
```

```python
Temperatures=[]
for line in lines :
  Temperature=line.split(',')[1]
  Temperatures.append(float(Temperature))  
print(Temperatures)
avg_temp=sum(Temperatures)/len(Temperatures)
print('avg_temp:',avg_temp)
```

#### The average maximum, minimum temperature, and total precipitation for a given period

```python
Max = 0
Min = 0
Prec = 0
Count = 0
rainy_days = 0
highest = 0
lowest = 0

for line in lines :  
  max_Temp=line.split(',')[1]
  min_Temp=line.split(',')[2]
  prec_Temp=line.split(',')[3]
  
  Max += float(max_Temp)
  Min += float(min_Temp)
  Prec += float(prec_Temp)
  Count += 1
  
  
  if float(prec_Temp) > 0:
     rainy_days+=1

  if float(max_Temp) > highest:
        highest = float(max_Temp)

  if float(min_Temp) < lowest:
        lowest = float(min_Temp)


  
maximum = Max/Count
maximum = Min/Count



print ('avg_temp_maximum:', maximum)
print ('avg_temp_maximum:', maximum)
print("Total Rain:", Prec)
print("Rainy Days:", rainy_days)
print("Highest_Temp:", highest)
print("Lowest_Temp:", lowest)
   

```
