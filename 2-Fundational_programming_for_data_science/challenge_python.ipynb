```python
import random 

# generate  random number int
number= random.randint(1,10)
count=0
while True:
    guess= input("Guess a number between 1 and 9 :")
    if guess.lower() == "exit":
      print("Thanks for playing")
      print(count)
      break
  
    count+=1
    guess= int(guess)
    if guess>number:
      print("You guess too high")
    elif guess<number:
      print("You guess too low")

    else:
      print("You guess right!")  
   
```

```python
import random
import string

Ltr=string.ascii_letters
Upper_ltr=string.ascii_uppercase
Lower_ltr=string.ascii_lowercase
punctuation=string.punctuation
number=string.digits

lis = ["sara","password","letter"]
password = random.choice(lis)
```

```python
print(Ltr)
print(Upper_ltr)
print(Lower_ltr)
print(punctuation)
print(number)
print(password)
```

```python
def PASSWORD(lenght,strenght):
    
    lis = ["Ihssane","password","letter",'sara']
    if strenght.lower() == 'weak':  
      return random.choice(lis)
      
    else:
      all_chars = Ltr + punctuation + number + Upper_ltr + Lower_ltr
      password = ""
      for i in range(lenght):  
          password += random.choice(all_chars)
      return password
    
```

```python
PASSWORD(10,'stronge')
```

```python
PASSWORD(10,'weak')
```
