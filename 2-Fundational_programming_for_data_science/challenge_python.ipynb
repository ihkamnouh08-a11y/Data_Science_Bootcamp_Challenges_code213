```python
def PASSWORD(lenght,strenght):
    
    lis = ["Ihssane","password","letter",'sara']
    if strenght.lower() == 'weak':  
      return random.choice(lis)
      
    else:
      all_chars = Ltr + punctuation + number + Upper_ltr + Lower_ltr
      password = ""
      for i in range(lenght):  
          password += random.choice(all_chars)
      return password
    
```

```python
import random
import string

def generate_password(length,strength):

    lst=("password","name","123456")
    if strength.lower() == "weak":
        return random.choice(lst)
    else:
        length=int(length)
        characters = string.ascii_letters + string.digits + string.punctuation
        password = ''.join(random.choice(characters) for i in range(length))
        return password

generate_password(10,"strong")
```
